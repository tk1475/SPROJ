```python
import osmnx as ox
import glob
import geopandas as gpd
import os

# Create output directories if they don't exist
os.makedirs('graphs', exist_ok=True)

# Prepare a list to collect stats
stats_list = []

# Iterate over all geojson files in societies/
for filepath in glob.glob('societies/*.geojson'):
    society_name = os.path.splitext(os.path.basename(filepath))[0]
    # Read the geojson as a GeoDataFrame
    gdf = gpd.read_file(filepath)
    # Get the convex hull of all geometries as the area boundary
    area = gdf.unary_union.convex_hull
    # Download the road network graph for the area
    G = ox.graph_from_polygon(area, network_type='drive')
    # Save the graph to file
    graph_path = f'graphs/{society_name}_graph.graphml'
    ox.save_graphml(G, graph_path)
    # Get basic stats
    stats = ox.basic_stats(G)
    stats['society'] = society_name
    stats_list.append(stats)

# Convert stats to DataFrame and save as CSV
stats_df = pd.DataFrame(stats_list)
stats_df.to_csv('road_stats.csv', index=False)
```

```python
# Improved Folium visualization: fixes missing societies by
# 1) Recursive search (subfolders) 2) Name normalization 3) Left join (show polygons even without stats)
# 4) Reports unmatched names 5) Single GeoJson per layer (faster)

import os, glob, math, re
import pandas as pd
import geopandas as gpd
import folium, branca
from pathlib import Path

# ---------------- Config ----------------
SOCIETIES_GLOB = 'societies/**/*.geojson'   # recursive
GRAPH_STATS_CSV = 'graph_stats.csv'
OUTPUT_HTML = 'society_network_metrics.html'
METRICS = [
    'edge_length_avg',
    'k_avg',
    'streets_per_node_avg',
    'street_length_avg',
    'circuity_avg'
]
WEIGHTS = {
    'edge_length_avg': 1.0,
    'k_avg': 1.0,
    'streets_per_node_avg': 1.2,
    'street_length_avg': 0.8,
    'circuity_avg': 0.6
}
LOWER_BETTER = {'circuity_avg': True}

def normalize_name(s: str) -> str:
    if not isinstance(s, str):
        return ''
    s = s.lower().strip()
    s = re.sub(r'\.geojson$', '', s)
    s = re.sub(r'[_\-\s]+', ' ', s)
    s = re.sub(r'[^a-z0-9 ]+', '', s)
    s = re.sub(r'\s+', ' ', s).strip()
    return s

# ---------------- Load Stats ----------------
if not Path(GRAPH_STATS_CSV).exists():
    raise FileNotFoundError(f"Missing {GRAPH_STATS_CSV}")
graph_stats_df = pd.read_csv(GRAPH_STATS_CSV)
graph_stats_df['society_norm'] = graph_stats_df['society'].map(normalize_name)

# ---------------- Load Polygons ----------------
poly_frames = []
for filepath in glob.glob(SOCIETIES_GLOB, recursive=True):
    try:
        gdf = gpd.read_file(filepath)
        if gdf.empty:
            continue
        if gdf.crs is None:
            gdf = gdf.set_crs('EPSG:4326')
        else:
            gdf = gdf.to_crs('EPSG:4326')
        society_raw = os.path.splitext(os.path.basename(filepath))[0]
        gdf['society_raw'] = society_raw
        gdf['society_norm'] = normalize_name(society_raw)
        poly_frames.append(gdf[['society_raw','society_norm','geometry']])
    except Exception as e:
        print(f"Skip {filepath}: {e}")

if not poly_frames:
    raise RuntimeError("No polygon GeoJSONs found.")

polygons_gdf = gpd.GeoDataFrame(pd.concat(poly_frames, ignore_index=True), crs='EPSG:4326')
# Dissolve multiparts per normalized name
polygons_gdf = polygons_gdf.dissolve(by='society_norm', as_index=False)

# ---------------- Merge (left join keeps polygons without stats) ----------------
merged = polygons_gdf.merge(graph_stats_df, on='society_norm', how='left', suffixes=('','_stat'))
merged['society_display'] = merged['society'].fillna(merged['society_norm'])

missing_stats = merged[merged['society'].isna()]
if not missing_stats.empty:
    print(f"Societies with no stats ({len(missing_stats)}):")
    print(", ".join(sorted(missing_stats['society_norm'].tolist())[:25]) + (" ..." if len(missing_stats)>25 else ""))

# ---------------- Metrics & Composite ----------------
for m in METRICS:
    if m not in merged.columns:
        print(f"Metric {m} missing in stats.")
        continue
    series = merged[m].astype(float)
    mu = series.mean(skipna=True)
    sd = series.std(ddof=0, skipna=True)
    if sd == 0 or pd.isna(sd):
        merged[f'{m}_z'] = 0.0
    else:
        z = (series - mu) / sd
        if LOWER_BETTER.get(m, False):
            z = -z
        merged[f'{m}_z'] = z.fillna(0)

def composite(row):
    total = 0.0
    wsum = 0.0
    for m,w in WEIGHTS.items():
        zc = f'{m}_z'
        if zc in row and pd.notna(row[zc]):
            total += row[zc]*w
            wsum += w
    return total/wsum if wsum else 0.0

merged['composite_score'] = merged.apply(composite, axis=1)
cmin, cmax = merged['composite_score'].min(), merged['composite_score'].max()
merged['composite_scaled'] = 0.5 if math.isclose(cmin,cmax) else (merged['composite_score']-cmin)/(cmax-cmin)

# ---------------- Folium Map ----------------
centroid = merged.geometry.unary_union.centroid
base_map = folium.Map(location=[centroid.y, centroid.x], zoom_start=12, tiles='cartodbpositron')

def make_colormap(series, palette='YlOrRd'):
    vals = series.dropna()
    if vals.empty:
        vals = pd.Series([0,1])
    vmin, vmax = vals.min(), vals.max()
    if math.isclose(vmin, vmax):
        vmax = vmin + 1e-6
    linear = getattr(branca.colormap.linear, f'{palette}_09', branca.colormap.linear.YlOrRd_09)
    return linear.scale(vmin, vmax)

def add_metric_layer(metric, gdf):
    if metric not in gdf.columns:
        return
    layer = folium.FeatureGroup(name=metric, show=False)
    cmap = make_colormap(gdf[metric])
    style_cache = {}
    def style_fn(feat):
        key = feat['properties']['society_display']
        if key in style_cache:
            return style_cache[key]
        val = feat['properties'].get(metric)
        color = '#cccccc' if pd.isna(val) else cmap(val)
        st = {'fillColor': color,'color':'#333','weight':0.7,'fillOpacity':0.6}
        style_cache[key] = st
        return st
    gj = folium.GeoJson(
        gdf[['society_display',metric,'geometry']].to_json(),
        style_function=style_fn,
        tooltip=folium.GeoJsonTooltip(
            fields=['society_display', metric],
            aliases=['Society', metric],
            localize=True
        ),
        highlight_function=lambda _: {'weight':2,'color':'#000'}
    )
    gj.add_to(layer)
    cmap.caption = metric
    layer.add_to(base_map)
    cmap.add_to(base_map)

for m in METRICS:
    add_metric_layer(m, merged)

# Composite (default visible)
comp_layer = folium.FeatureGroup(name='composite_score', show=True)
comp_cmap = make_colormap(merged['composite_score'])
def comp_style(feat):
    val = feat['properties'].get('composite_score')
    color = '#bbbbbb' if pd.isna(val) else comp_cmap(val)
    return {'fillColor': color,'color':'#111','weight':1,'fillOpacity':0.65}

folium.GeoJson(
    merged[['society_display','composite_score','geometry']].to_json(),
    style_function=comp_style,
    tooltip=folium.GeoJsonTooltip(
        fields=['society_display','composite_score'],
        aliases=['Society','Composite'],
        localize=True
    ),
    highlight_function=lambda _: {'weight':2,'color':'#000'}
).add_to(comp_layer)
comp_cmap.caption = 'composite_score'
comp_layer.add_to(base_map)
comp_cmap.add_to(base_map)

folium.LayerControl(collapsed=False).add_to(base_map)
base_map.save(OUTPUT_HTML)
base_map
```
